# Scope and Data Contract

This notebook is the entry point for the restricted code-inspection package supporting the Banjarbaru manuscript. It is intentionally Markdown-only and does not load data, execute models, or reproduce manuscript metrics.

## Purpose

Supports inspection of the final analytical workflow, methodological decisions, and privacy-reviewed aggregate outputs. It is not an independently runnable full-data reproduction package.

## Restricted Inputs

The Banjarbaru source dataset, processed reviews, business-level metadata, manual-audit records, row-level predictions, probability vectors, checkpoints, vocabularies, tokenizer caches, and bootstrap replicate files are intentionally excluded. The analysis notebooks may describe authorized inputs that remain outside this package.

## Analytical Contract

The manuscript workflow used LSTM-derived silver labels, minimal text preprocessing, an Actual Place Name exact-case grouping rule, repeated seeds 42, 52, and 62, five outer folds for OOF and ensemble prediction generation, five final models, manual-label evaluation on a stratified locked-test audit, paired statistical analysis, probability reporting, and fixed-budget triage.

No business overlap is claimed under the grouping rule used to create and evaluate the split. This package does not introduce a new grouping rule or modify the final analysis protocol.

## Notebook Sequence

The release sequence proceeds from LSTM silver-label generation, preprocessing, group-aware partitioning, descriptive checks, and provenance verification to repeated-seed base-model training, stacking, calibration, aggregation, manual-label evaluation, paired statistics, triage, probability reporting, feature ablation, and base-model complementarity.

Each later notebook should contain English Markdown describing its purpose, authorized inputs, aggregate outputs, and methodological boundaries before its code cells.

## Aggregate Outputs

Privacy-reviewed aggregate files are stored under `aggregate_outputs/` in five stage-specific directories: LSTM provenance, human-label audit, final-model statistics, triage and probability reporting, and diagnostics/descriptives/ablation/complementarity. Individual review records and model artifacts are not included.
